# 3. Explain the use of colour in the dashboard

| Gestalt principle | How is it applied in the dashboard? | Why? |
|---|---|---|
| Similarity | The same colour is used for all bars in the bar chart and for the line in the line chart. | It creates visual consistency and helps users understand that both charts represent accident data. |
| Figure-ground | A light background is used with dark-coloured bars, lines, and text. | It improves contrast and makes the charts easier to read. |
| Common region | Each chart is placed in a separate, clearly defined area of the dashboard. | It helps users distinguish between the two visualizations and understand the different information each one presents. |

## First draft of the dashboard

The dashboard will contain two charts:

- **Line chart:** Placed at the top, showing the number of accidents per year to identify trends over time.
- **Bar chart:** Placed below the line chart, showing the number of accidents per moment of the day to compare accident counts across time periods.

Both charts will use consistent colours, clear labels, readable axes, and a simple background to improve readability and make the information easy to understand.
